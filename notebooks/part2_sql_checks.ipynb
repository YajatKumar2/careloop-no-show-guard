```python
import pandas as pd

df = pd.read_csv("../outputs/appointments_clean.csv")
clinics = pd.read_csv("../outputs/clinics_enriched.csv")

m = df.merge(clinics, on="clinic_id")
g = m.groupby(["clinic_name", "city"]).agg(appointments=("no_show", "size"), no_shows=("no_show", "sum"))
g["no_show_pct"] = (100 * g["no_shows"] / g["appointments"]).round(1)
print(g.sort_values("no_show_pct", ascending=False))
```
